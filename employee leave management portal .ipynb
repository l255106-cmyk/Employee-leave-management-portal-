```python
import json
import os
from datetime import datetime
```

```python
employees = {
    "E1": {"name": "Muhammad Ali", "department": "IT", "password": "Ali123"},
    "E2": {"name": "Ayeza Khan", "department": "HR", "password": "Ayeza123"},
    "E3": {"name": "Ahsan Farhan", "department": "Finance", "password": "Ahsan123"},
    "E4": {"name": "Ayesha Javaid", "department": "Marketing", "password": "Ayesha123"},
    "E5": {"name": "Usman Tariq", "department": "Sales", "password": "Usman123"}
}

FILE_NAME = "employee_leave_data.json"
```

```python
def load_records():
    if not os.path.exists(FILE_NAME):
        return []

    try:
        with open(FILE_NAME, "r") as file:
            data = json.load(file)

        if not isinstance(data, list):
            return []

        records = []

        for record in data:
            if not isinstance(record, dict):
                continue

            if record.get("employee_id") not in employees:
                continue

            required_fields = [
                "name", "department", "start_date",
                "end_date", "days", "reason", "status"
            ]

            if not all(field in record for field in required_fields):
                continue

            if not isinstance(record["days"], int) or record["days"] < 1:
                continue

            records.append(record)

        return records

    except (json.JSONDecodeError, OSError):
        print("Could not read the saved records.")
        return []


def save_records(records):
    try:
        with open(FILE_NAME, "w") as file:
            json.dump(records, file, indent=4)

        return True

    except OSError:
        print("Could not save the records.")
        return False


leave_records = load_records()
```

```python
def user_login():
    employee_id = input("Enter Employee ID: ").strip().upper()
    password = input("Enter Password: ").strip()

    if employee_id in employees:
        if employees[employee_id]["password"] == password:
            print("\nLogin successful!")
            print("Welcome", employees[employee_id]["name"])
            return employee_id

    print("Invalid Employee ID or Password.")
    return None
```

```python
def add_record(employee_id):
    try:
        start_text = input("Enter start date (DD-MM-YYYY): ").strip()
        end_text = input("Enter end date (DD-MM-YYYY): ").strip()

        start_date = datetime.strptime(start_text, "%d-%m-%Y")
        end_date = datetime.strptime(end_text, "%d-%m-%Y")

        if end_date < start_date:
            print("End date cannot be before start date.")
            return

        reason = input("Enter reason for leave: ").strip()

        if reason == "":
            print("Please enter a reason for leave.")
            return

        days = (end_date - start_date).days + 1

        record = {
            "employee_id": employee_id,
            "name": employees[employee_id]["name"],
            "department": employees[employee_id]["department"],
            "start_date": start_date.strftime("%d-%m-%Y"),
            "end_date": end_date.strftime("%d-%m-%Y"),
            "days": days,
            "reason": reason,
            "status": "Leave Application Recorded"
        }

        leave_records.append(record)

        if save_records(leave_records):
            print("Leave application recorded successfully.")
            print("Total leave days:", days)
        else:
            leave_records.pop()

    except ValueError:
        print("Please enter valid dates in DD-MM-YYYY format.")
```

```python
def view_records(employee_id):
    found = False

    for record in leave_records:
        if record["employee_id"] == employee_id:
            found = True

            print("\nStart Date:", record["start_date"])
            print("End Date:", record["end_date"])
            print("Days:", record["days"])
            print("Reason:", record["reason"])
            print("Status:", record["status"])
            print("-------------------------")

    if not found:
        print("You have no leave records.")
```

```python
def search_record(employee_id):
    search_text = input(
        "Enter start date to search (DD-MM-YYYY): "
    ).strip()

    try:
        search_date = datetime.strptime(
            search_text, "%d-%m-%Y"
        ).strftime("%d-%m-%Y")

    except ValueError:
        print("Please enter a valid date in DD-MM-YYYY format.")
        return

    found = False

    for record in leave_records:
        if (record["employee_id"] == employee_id and
                record["start_date"] == search_date):

            found = True

            print("\nLeave Record Found")
            print("Start Date:", record["start_date"])
            print("End Date:", record["end_date"])
            print("Days:", record["days"])
            print("Reason:", record["reason"])
            print("Status:", record["status"])

    if not found:
        print("No leave record found for that date.")
```

```python
def generate_report(employee_id):
    total_applications = 0
    total_days = 0

    for record in leave_records:
        if record["employee_id"] == employee_id:
            total_applications += 1
            total_days += record["days"]

    print("\nLeave Report")
    print("Total Applications:", total_applications)
    print("Total Leave Days:", total_days)
```

```python
def employee_menu(employee_id):
    while True:
        print("\nEmployee Menu")
        print("1. Apply for Leave")
        print("2. View Leave Records")
        print("3. Search Leave Record")
        print("4. Generate Report")
        print("5. Logout")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            add_record(employee_id)

        elif choice == "2":
            view_records(employee_id)

        elif choice == "3":
            search_record(employee_id)

        elif choice == "4":
            generate_report(employee_id)

        elif choice == "5":
            print("Logged out successfully.")
            break

        else:
            print("Invalid choice. Please try again.")
```

```python
def main():
    while True:
        print("\nEmployee Leave Management Portal")
        print("1. Employee Login")
        print("2. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            employee_id = user_login()

            if employee_id is not None:
                employee_menu(employee_id)

        elif choice == "2":
            print("Thank you for using the portal.")
            break

        else:
            print("Invalid choice. Please try again.")


if __name__ == "__main__":
    main()
```
